# Data Understanding
In this section we will explore and analize the dataset, detect missing values and identify patterns that will guide the data preparation and modeling steps.

## Imports

```python
import pandas as pd
import numpy  as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Import the dataset

```python
ds = pd.read_csv("data/escs_trend.csv")
ds.head()
```

```python
print("Dataset shape:", ds.shape)
ds.info()
```

```python
ds.describe()
```

```python
ds.isnull().sum()
```

```python
# correlation matrix of categorical features
categorical_features = ds.select_dtypes(include=['int','float']).columns
correlation_matrix = ds[categorical_features].apply(lambda x: pd.factorize(x)[0]).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Categorical Features')
plt.show()
```
