### Students Dataset

```python
# filtrar Portugal
df_fl2_pt = dt_fl2[dt_fl2['CNT'] == 'PRT'].copy()
df_fl2_pt = df_fl2_pt[["CNTSTUID"] + col_dt_fl2]

df_std_pt = df_fl2_pt.copy()


# 2. ACRESCENTAR dt_stdc (SÓ NOVOS ALUNOS)

# filtrar Portugal
df_stdc_pt = dt_stdc[dt_stdc['CNT'] == 'PRT'].copy()

# garantir mesmas colunas
df_stdc_pt = df_stdc_pt.reindex(columns=df_std_pt.columns)


# encontrar alunos que ainda não existem
mask_novos = ~df_stdc_pt["CNTSTUID"].isin(df_std_pt["CNTSTUID"])
df_novos = df_stdc_pt[mask_novos]

# acrescentar ao dataset
df_std_pt = pd.concat([df_std_pt, df_novos], ignore_index=True)

# 3. ACRESCENTAR RESULTADOS DOS TESTES (SÓ ALUNOS EXISTENTES)
df_ct = dt_ct[dt_ct['CNT'] == 'PRT'].copy()
df_ct = df_ct[["CNTSTUID"] + col_dt_ct]
df_std_pt = pd.merge(
    df_std_pt,
    df_ct,
    on="CNTSTUID",
     how="left"   # só acrescenta resultados aos alunos que já existem
)

# 4. dados por escola
df_shc_pt = dt_shc[dt_shc['CNT'] == 'PRT'].copy()
df_shc_pt = df_shc_pt[["CNTSCHID"] + col_dt_shc]

df_std_pt = pd.merge(
    df_std_pt,
    df_shc_pt,
    on="CNTSCHID",
    how="left"
)

# 5. dados dos profs
df_tc_pt = dt_tq[dt_tq['CNT'] == 'PRT'].copy()

prof_agg = df_tc_pt.groupby("CNTSCHID").agg({
    "TC001Q01NA": "mean",                       # género médio
    "TC002Q01NA": "mean",                       # idade média
    "TC007Q02NA": "mean",                       # experiência média
    "EMPLSTAT":   lambda x: (x == 1).mean(),    # % permanentes
}).reset_index()

prof_agg.columns = [
    "CNTSCHID",
    "TC_GENERO_MEDIA",
    "TC_IDADE_MEDIA",
    "TC_EXP_MEDIA",
    "TC_PERM_PERC",
]

df_std_pt = pd.merge(
    df_std_pt,
    prof_agg,
    on="CNTSCHID",
    how="left"
)

```
